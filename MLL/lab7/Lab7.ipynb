#### 1. Implement in python program of the following problems using Bayes Theorem. 

Of the students in the college, 60% of the students reside in the hostel and 40% of the students are day scholars. Previous year results report that 30% of all students who stay in the hostel scored A Grade and 20% of day scholars scored A grade. At the end of the year, one student is chosen at random and found that he/she has an A grade. What is the probability that the student is a hosteler?

```python
p_h = 0.6       # hosteler
p_d = 0.4       # day scholar

p_a_h = 0.3     # A grade given hosteler
p_a_d = 0.2     # A grade given day scholar

# total probability of getting A grade
p_a = (p_a_h * p_h) + (p_a_d * p_d)

p_h_a = (p_a_h * p_h) / p_a

print(f"Probability of A grade student being a hosteler: {p_h_a}")
```

Suppose you're testing for a rare disease, and you have the following information: 
- The disease has a prevalence of 0.01 (1% of the population has the disease). 
- The test is not perfect:
    - The test correctly identifies the disease (true positive) 99% of the time (sensitivity). 
    - The test incorrectly indicates the disease (false positive) 2% of the time (1 - specificity).

Calculate the probability of having the disease given a positive test result using Bayes' theorem.

```python
p_d = 0.01          # having disease
p_nd = 1 - p_d      # not having disease

p_pos_d = 0.99      # TPR
p_pos_nd = 0.02     # FPR

# total probability of positive test
p_pos = (
    p_pos_d * p_d
    + p_pos_nd * p_nd
)

p_d_pos = (p_pos_d * p_d) / p_pos

print(f"Probability of having the disease given a positive test: {p_d_pos}")
```

#### 2. Develop a function python code for Naïve Bayes classifier from scratch without using scikit-learn library, to predict whether the buyer should buy computer or not. Consider a following sample training dataset stored in a CSV file containing information about following buyer conditions (such as “<=30,” “medium,” “Yes,” and “fair”) and whether the player played golf (“Yes” or “No”).

```python
data = {
    "age": [
        "<=30", "<=30", "31...40", ">40", ">40", ">40", "31...40",
        "<=30", "<=30", ">40", "<=30", "31...40", "31...40", ">40"
    ],
    
    "income": [
        "high", "high", "high", "medium", "low", "low", "low",
        "medium", "low", "medium", "medium", "medium", "high", "medium"
    ],
    
    "student": [
        "no", "no", "no", "no", "yes", "yes", "yes",
        "no", "yes", "yes", "yes", "no", "yes", "no"
    ],
    
    "credit_rating": [
        "fair", "excellent", "fair", "fair", "fair", "excellent", "excellent",
        "fair", "fair", "fair", "excellent", "excellent", "fair", "excellent"
    ],
    
    "buys_computer": [
        "no", "no", "yes", "yes", "yes", "no", "yes",
        "no", "yes", "yes", "yes", "yes", "yes", "no"
    ]
}

df = pd.DataFrame(data)
df.to_csv("computer.csv", index=False)
```

```python
df = pd.read_csv("computer.csv")
df.head()
```

```python
def naive_bayes_predict(df, sample, target="buys_computer"):
    classes = df[target].unique()
    features = df.columns.drop(target)

    probabilities = {}

    for cls in classes:
        class_data = df[df[target] == cls]

        prior_probability = len(class_data) / len(df)
        probability = prior_probability

        for feature in features:
            value = sample[feature]
            count = len(class_data[class_data[feature] == value])
            conditional_probability = count / len(class_data)
            probability *= conditional_probability

        probabilities[cls] = probability

    predicted_class = max(probabilities, key=probabilities.get)

    return predicted_class, probabilities
```

```python
sample = {
    "age": "<=30",
    "income": "medium",
    "student": "yes",
    "credit_rating": "fair"
}

predicted, probabilities = naive_bayes_predict(df, sample)

print("Probabilities:")
for cls, prob in probabilities.items():
    print(f"{cls}: {prob}")

print(f"\nPredicted class: {predicted}")
```

#### 3. Write a Python function to implement the Naive Bayes classifier without using the scikit-learn library for the following sample training dataset stored as a .CSV file.
- Build a classifier that determines whether a text is about sports or not. 
- Determine which tag the sentence "A very close game" belongs to.

```python
data = {
    "text": [
        "A great game",
        "The election was over",
        "Very clean match",
        "A clean but forgettable game",
        "It was a close election"
    ],

    "tag": [
        "Sports",
        "Not Sports",
        "Sports",
        "Sports",
        "Not Sports"
    ]
}

df = pd.DataFrame(data)
df.to_csv("sports.csv", index=False)
```

```python
df = pd.read_csv("sports.csv")
df.head()
```

```python
import re
def tokenize(text):
    return re.findall(r'\b\w+\b', text.lower())
```

```python
from collections import Counter

def naive_bayes_train(df, target='tag'):
    classes = df[target].unique()

    class_counts = {}
    word_counts = {}
    total_words = {}
    vocabulary = set()

    for cls in classes:
        class_data = df[df[target] == cls]
        class_counts[cls] = len(class_data)

        words = []

        for text in class_data['text']:
            tokens = tokenize(text)
            words.extend(tokens)
            vocabulary.update(tokens)

        word_counts[cls] = Counter(words)
        total_words[cls] = len(words)

    return class_counts, word_counts, total_words, vocabulary
```

```python
class_counts, word_counts, total_words, vocabulary = naive_bayes_train(df)

print(f"Class counts: {class_counts}")
print(f"\nWord counts: {word_counts}")
print(f"\nVocabulary: {vocabulary}")
```

```python
def naive_bayes_predict(df, text):
    class_counts, word_counts, total_words, vocabulary = naive_bayes_train(df)

    words = tokenize(text)

    total_documents = len(df)

    probabilities = {}

    for cls in class_counts:
        prior_probability = class_counts[cls] / total_documents
        probability = prior_probability

        for word in words:
            count = word_counts[cls][word]

            word_probability = (
                (count + 1) /
                (total_words[cls] + len(vocabulary))
            )

            probability *= word_probability

        probabilities[cls] = probability

    predicted_class = max(probabilities, key=probabilities.get)

    return predicted_class, probabilities
```

```python
sentence = "A very close game"

predicted, probabilities = naive_bayes_predict(df, sentence)

print("Probabilities:")
for cls, prob in probabilities.items():
    print(f"{cls}: {prob}")

print(f"\nPredicted class: {predicted}")
```

#### 2. Develop a function python code for Naïve Bayes classifier from scratch without using scikit-learn library, to predict whether the player should play golf based on weather conditions. Consider a following sample training dataset stored in a CSV file containing information about following weather conditions (such as “Sunny,” “Hot,” “High Humidity,” and “Not Windy”) and whether the player played golf (“Yes” or “No”).

```python
data = {
    "weather": [
        "Rainy", "Rainy", "Overcast", "Sunny", "Sunny", "Sunny",
        "Overcast", "Rainy", "Rainy", "Sunny", "Rainy",
        "Overcast", "Overcast", "Sunny"
    ],

    "temperature": [
        "Hot", "Hot", "Hot", "Mild", "Cool", "Cool",
        "Cool", "Mild", "Cool", "Mild", "Mild",
        "Mild", "Hot", "Mild"
    ],

    "humidity": [
        "High", "High", "High", "High", "Normal", "Normal",
        "Normal", "High", "Normal", "Normal", "Normal",
        "High", "Normal", "High"
    ],

    "windy": [
        False, True, False, False, False, True,
        True, False, False, False, True,
        True, False, True
    ],

    "play": [
        "No", "No", "Yes", "Yes", "Yes", "No",
        "Yes", "No", "Yes", "Yes", "Yes",
        "Yes", "Yes", "No"
    ]
}

df = pd.DataFrame(data)
df.to_csv("golf.csv", index=False)
```

```python
df = pd.read_csv("golf.csv")
df.head()
```

```python
def naive_bayes_predict(df, sample, target='play'):
    classes = df[target].unique()

    probabilities = {}

    for cls in classes:
        class_data = df[df[target] == cls]
        prior_probability = len(class_data) / len(df)

        probability = prior_probability

        for feature, value in sample.items():
            count = len(class_data[class_data[feature] == value])
            possible_values = df[feature].unique()

            conditional_probability = (
                count + 1
            ) / (
                len(class_data) + len(possible_values)
            )

            probability *= conditional_probability

        probabilities[cls] = probability

    predicted_class = max(probabilities, key=probabilities.get)
    return predicted_class, probabilities
```

#### 3. Write a function Python program to implement the naïve Bayesian classifier without using scikit-learn library for a following sample training data set stored as a.CSV file. Calculate the accuracy, precision, and recall for your train/test data set.

```python
data = {
    "text": [
        "I love this sandwich",
        "This is an amazing place",
        "I feel very good about these beers",
        "This is my best work",
        "What an awesome view",
        "I do not like this restaurant",
        "I am tired of this stuff",
        "I can't deal with this",
        "He is my sworn enemy",
        "My boss is horrible",
        "This is an awesome place",
        "I do not like the taste of this juice",
        "I love to dance",
        "I am sick and tired of this place",
        "What a great holiday",
        "That is a bad locality to stay",
        "We will have good fun tomorrow",
        "I went to my enemy's house today"
    ],

    "label": [
        "pos", "pos", "pos", "pos", "pos",
        "neg", "neg", "neg", "neg", "neg",
        "pos", "neg", "pos", "neg", "pos",
        "neg", "pos", "neg"
    ]
}

df = pd.DataFrame(data)
df.to_csv("sentiment.csv", index=False)
```

```python
df = pd.read_csv("sentiment.csv")
df.head()
```

```python
df = df.sample(frac=1, random_state=67).reset_index(drop=True)
split = int(0.8 * len(df))

train = df.iloc[:split]
test = df.iloc[split:]
```

```python
import re

def tokenize(text):
    return re.findall(r'\b\w+\b', text.lower())
```

```python
from collections import Counter

def naive_bayes_train(train):
    classes = train['label'].unique()

    class_counts = {}
    word_counts = {}
    total_words = {}
    vocabulary = set()

    for cls in classes:
        class_data = train[train['label'] == cls]
        class_counts[cls] = len(class_data)
        words = []

        for text in class_data['text']:
            tokens = tokenize(text)

            words.extend(tokens)
            vocabulary.update(tokens)

        word_counts[cls] = Counter(words)
        total_words[cls] = len(words)

    return class_counts, word_counts, total_words, vocabulary
```

```python
import math

def naive_bayes_predict(text):
    class_counts, word_counts, total_words, vocabulary = naive_bayes_train(train)

    words = tokenize(text)
    probabilities = {}

    for cls in class_counts:
        prior_probability = class_counts[cls] / len(train)
        probability = math.log(prior_probability)

        for word in words:
            count = word_counts[cls][word]

            word_probability = (
                count + 1
            ) / (
                total_words[cls] + len(vocabulary)
            )

            probability += math.log(word_probability)

        probabilities[cls] = probability

    return max(probabilities, key=probabilities.get)
```

```python
actual = []
predicted = []

for _, row in test.iterrows():
    result = naive_bayes_predict(row['text'])

    actual.append(row['label'])
    predicted.append(result)

print(f"Actual classes: {actual}")
print(f"Predicted classes: {predicted}")
```

```python
tp = tn = fp = fn = 0

for actual_value, predicted_value in zip(actual, predicted):
    if actual_value == "pos" and predicted_value == "pos":
        tp += 1

    elif actual_value == "neg" and predicted_value == "neg":
        tn += 1

    elif actual_value == "pos" and predicted_value == "neg":
        fn += 1

    elif actual_value == "neg" and predicted_value == "pos":
        fp += 1

print(f"True positives: {tp}")
print(f"True negatives: {tn}")
print(f"False positives: {fp}")
print(f"False negatives: {fn}")
```

```python
accuracy = (tp + tn) / (tp + tn + fp + fn)
precision = tp / (tp + fp) if (tp + fp) != 0 else 0
recall = tp / (tp + fn) if (tp + fn) != 0 else 0

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
```
